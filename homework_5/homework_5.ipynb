```python
import json
import csv

purchases = {}

with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    f.readline()

    for line in f:
        purchase = json.loads(line)
        purchases[purchase['user_id']] = purchase['category']

print('Количество покупателей:', len(purchases))
```

```python
with open('visit_log.csv', 'r', encoding='utf-8', newline='') as visits_file, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel_file:

    reader = csv.DictReader(visits_file)

    writer = csv.DictWriter(
        funnel_file,
        fieldnames=['user_id', 'source', 'category']
    )
    writer.writeheader()

    for visit in reader:
        user_id = visit['user_id']

        if user_id in purchases:
            writer.writerow({
                'user_id': user_id,
                'source': visit['source'],
                'category': purchases[user_id]
            })
```
