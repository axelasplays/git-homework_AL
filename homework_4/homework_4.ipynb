```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}


def find_document_owner(doc_number):

    for document in documents:
        if document['number'] == doc_number:
            return document['name']

    return None


def find_document_shelf(doc_number):

    for shelf, numbers in directories.items():
        if doc_number in numbers:
            return shelf

    return None


def process_command(command):

    if command == 'p':
        doc_number = input('Введите номер документа:\n')
        owner = find_document_owner(doc_number)

        if owner:
            print(f'Владелец документа: {owner}')
        else:
            print('Владелец документа: владелец не найден')

    elif command == 's':
        doc_number = input('Введите номер документа:\n')
        shelf = find_document_shelf(doc_number)

        if shelf:
            print(f'Документ хранится на полке: {shelf}')
        else:
            print('Документ не найден')

    elif command == 'q':
        return False

    else:
        print('Команда не найдена')

    return True


def main():

    while True:
        command = input('Введите команду:\n')

        if not process_command(command):
            print('Программа завершена')
            break


main()
```
