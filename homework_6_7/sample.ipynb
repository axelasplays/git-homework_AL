```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
def load_data(file_path, limit=15):
    data = pd.read_csv(file_path)

    data = data.head(limit)

    data['age'] = pd.to_numeric(data['age'], errors='coerce')
    data['bill'] = pd.to_numeric(data['bill'], errors='coerce')

    data = data.dropna(subset=['age', 'bill'])

    return data
```

```python
def plot_seaborn_scatter(data):
    sns.scatterplot(
        data=data,
        x='age',
        y='bill'
    )

    plt.xlabel('age')
    plt.ylabel('bill')

    plt.show()
```

```python
def plot_plotly_scatter(data):
    fig = px.scatter(
        data,
        x='age',
        y='bill',
        labels={
            'age': 'age',
            'bill': 'bill'
        }
    )

    fig.show()
```

```python
def plot_regression(data):
    sns.regplot(
        data=data,
        x='age',
        y='bill'
    )

    plt.xlabel('age')
    plt.ylabel('bill')

    plt.show()
```

```python
def main(file_path):
    data = load_data(file_path)

    print("Исходные данные:")
    print(data[['age', 'bill']])

    print("Seaborn")
    plot_seaborn_scatter(data)

    print("Plotly")
    plot_plotly_scatter(data)
```

```python
main("web_clients_correct.csv")
```
