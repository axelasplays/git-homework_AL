```python
import pandas as pd
```

```python
def load_data(file_path, limit=15):
    data = pd.read_csv(file_path)

    data = data.head(limit)

    data['age'] = pd.to_numeric(data['age'], errors='coerce')

    data = data.dropna(subset=['age'])

    return data
```

```python
def analyze_age(age):
    if age > 25:
        return 'Старше 25'
    else:
        return 'Младше или равен 25'
```

```python
def process_age(data):
    result = data['age'].apply(analyze_age)

    return result
```

```python
def task_1(file_path):
    data = load_data(file_path)

    result = process_age(data)

    print(result)
```

```python
task_1("web_clients_correct.csv")
```

```python
def filter_male_clients(data):
    result = data[
        (data['sex'] == 'male') &
        (data['age'] > 30)
    ]

    return result
```

```python
def task_2(file_path):
    data = load_data(file_path)

    result = filter_male_clients(data)

    print(result)
```

```python
task_2("web_clients_correct.csv")
```
